# Imports

```python
from pathlib import Path
import pandas as pd
from diive.core.io.files import save_parquet, load_parquet
```

# Load L3.3 results for NEE, LE and H

```python
nee = load_parquet(filepath=r"41_NEE/41.1_FluxProcessingChain_L3.2_NEE.parquet")
le = load_parquet(filepath=r"42_LE/42.1_FluxProcessingChain_L3.2_LE.parquet")
h = load_parquet(filepath=r"43_H/43.1_FluxProcessingChain_L3.2_H.parquet")
n2o = load_parquet(filepath=r"44_FN2O/44.1_FluxProcessingChain_L3.2_FN2O.parquet")
ch4 = load_parquet(filepath=r"45_FCH4/45.1_FluxProcessingChain_L3.2_FCH4.parquet")
```

```python
# Load METEO
SOURCEDIR = r"../10_METEO"
FILENAME = r"13.1_CH-TAN_meteo_gapfilled.parquet"
FILEPATH = Path(SOURCEDIR) / FILENAME
meteo = load_parquet(filepath=FILEPATH)
```

# Load main data

```python
SOURCEDIR = r"../30_MERGE_DATA"
FILENAME = r"33.1_CH-TAN_IRGA+LGR_Level-1_eddypro_fluxnet.parquet"
FILEPATH = Path(SOURCEDIR) / FILENAME
maindf = load_parquet(filepath=FILEPATH)
```

## Add results for NEE

```python
newcols = [c for c in nee.columns if c not in maindf]
print("NEW VARIABLES FROM FLUX PROCESSING CHAIN:")
[print(f"++ {c}") for c in newcols]
maindf = pd.concat([maindf, nee[newcols]], axis=1)
maindf
```

# Export 

## Export all data

```python
filename = "46.1_FLUXES_L3.2_NEE_LE_H_FN2O_FCH4"
save_parquet(data=maindf, filename=filename)
```

## Export subset for REddyProc

```python
from diive.core.times.times import insert_timestamp
export_REddyProc = maindf.copy()
```

```python
newcols = [c for c in meteo.columns if c not in export_REddyProc]
export_REddyProc = pd.concat([export_REddyProc, meteo[newcols]], axis=1)
```

```python
exportcols = ['NEE_L3.1_L3.2_QCF', 'LE_L3.1_L3.2_QCF', 'H_L3.1_L3.2_QCF',
              'USTAR',
              'sw_in', 'ta', 'vpd']
export_REddyProc = export_REddyProc[exportcols].copy()
locs = (export_REddyProc.index.year >= 2024) & (export_REddyProc.index.year <= 2025)
export_REddyProc = export_REddyProc[locs].copy()
export_REddyProc = insert_timestamp(data=export_REddyProc, convention="end", set_as_index=True)
export_REddyProc
```

```python
filename = "46.2_FluxProcessingChain_L3.2_subset-forREddyProc"
export_REddyProc.to_csv(f"{filename}.csv", index=True)
```
